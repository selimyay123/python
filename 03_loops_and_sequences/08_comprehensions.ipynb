```python
even_numbers = []

for num in range(21):
    if num % 2 == 0:
        even_numbers.append(num)

print(even_numbers)

# There is a more concise way to write this that uses list comprehension instead

# List comprehension allows you to create a new list in a single line by combining a loop and condition directly within square brackets

# SYNTAX: OUTPUT | LOOP | CONDITION
# num => output
# for num in range(21) => loop
# if num % 2 == 0 => condition

even_numbers = [num for num in range(21) if num % 2 == 0]
print(even_numbers)

numbers = [1, 2, 3, 4, 5]
result = [(num, 'Even') if num % 2 == 0 else (num, 'Odd') for num in numbers]
print(result)
```

```python
# filter()

## The filter() function is used to select elements from an iterable that meet a specific condition. The filter() function accepts a function and an iterable for its arguments

words = ['tree', 'sky', 'mountain', 'river', 'cloud', 'sun']

def is_long_word(word):
    return len(word) > 4

long_words = list(filter(is_long_word, words))
print(long_words)
```

```python
# map() 

## takes an iterable and applies a function to each of its elements

celsius = [0, 10, 20, 30, 40]

def to_fahrenheit(temp):
    return (temp * 9/5) + 32

fahrenheit = list(map(to_fahrenheit, celsius))
```

```python
# sum()

## This function is used to get the sum from an iterable like a list or tuple

numbers = [5, 10, 15, 20]
total = sum(numbers)
print(total)

## You can also pass in an optional start argument which sets the initial value for the summation

numbers = [5, 10, 15, 20]
total = sum(numbers, 10)
print(total)

## You can also choose to use the start argument as a keyword argument like this instead:

numbers = [5, 10, 15, 20]
total = sum(numbers, start=10)
print(total)
```
