```python
# append()

## adds an item to the end of the list
numbers = [1, 2, 3, 4, 5]
numbers.append(6)
print('1:', numbers)
# you can also add one list at the end of another
even_numbers = [6, 8, 10]
numbers.append(even_numbers)
print('2:', numbers)

# extend()

## it adds all of the individual elements from one list at the end of another
numbers = [1, 2, 3, 4, 5]
numbers.extend(even_numbers)
print('3:', numbers)

# insert()

## it inserts an element at a specific index in a list
## This method accepts two arguments: the index where you wish to insert the new item and the item you want to insert
numbers = [1, 2, 3, 4, 5]
numbers.insert(2, 2.5)
print('4:', numbers)

# remove()

## removes an element from a list
## The remove() method takes the value of the element to remove as an argument
## It is important to note that this method will only remove the first occurrence of an item. Not all of them
numbers = [10, 20, 30, 40, 50, 50]
numbers.remove(50)
print('5:', numbers)

# pop()

## removes an element at a specific index
## If you don't specify an element for the pop method, then the last element is removed.
numbers = [1, 2, 3, 4, 5]
numbers.pop(1) # The number 2 is returned

# clear()

## empty the list
numbers = [1, 2, 3, 4, 5]
numbers.clear()
print('6:', numbers)

# sort()

numbers = [19, 2, 35, 1, 67, 41]
numbers.sort()
print('7:', numbers)

# sorted()

## n contrast to the sort() method, there is the sorted() function which works for any iterable and returns a new sorted list instead of modifying the original list
numbers = [19, 2, 35, 1, 67, 41]
sorted_numbers = sorted(numbers)
print('8.1:', numbers)
print('8.2:', sorted_numbers)
## Both the sort() method and sorted() function accept optional key and reverse parameters.

# reverse()

numbers = [6, 5, 4, 3, 2, 1]
numbers.reverse()
print('9:', numbers)

# index()

## finds the first index where an element can be found in a list
programming_languages = ['Rust', 'Java', 'Python', 'C++']
print('10:', programming_languages.index('Java'))
## If the element cannot be found, then Python throws a ValueError:
```
