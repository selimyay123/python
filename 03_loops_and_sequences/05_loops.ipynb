```python
# loops are used to repeat a block of code for a set number of times

## for loop
programming_languages = ['Rust', 'Java', 'Python', 'C++']
for language in programming_languages:
    print(language)

## You can also use a for loop to iterate through other iterables like a string
for char in 'code':
    print(char)

## you can also nest for loops in Python
## The outer loop will iterate through each category in the categories list. For each category, the inner loop will iterate through each food in the foods list

categories = ['Fruit', 'Vegetable']
foods = ['Apple', 'Carrot', 'Banana']

for category in categories:
    for food in foods:
        print(category, food)
```

```python
# while loop

## loop will repeat a block of code until the condition is False
secret_number = 3
guess = 0

while guess != secret_number:
    guess = int(input('Guess the number (1-5): '))
    if guess != secret_number:
        print('Wrong! Try again.')
print('You got it!')
```

```python
# break and continue

## Just like in JavaScript, Python supports the break and continue statements
## The break statement is used to stop the execution of a loop
## In this example, we iterate through a list of developer_names and print each name to the console. If the name is equal to Naomi, then we break out of the loop. This results in only the name Jess being printed to the console
developer_names = ['Jess', 'Naomi', 'Tom']

for developer in developer_names:
    if developer == 'Naomi':
        break
    print(developer)

## The continue statement is used to skip the current iteration of a loop and move onto the next iteration
## Now the result in the console will be different. The names Jess and Tom are printed because the continue statement skips the second iteration of the loop when developer is equal to Naomi, and does not print that name to the console
developer_names = ['Jess', 'Naomi', 'Tom']

for developer in developer_names:
    if developer == 'Naomi':
        continue
    print(developer)

## Both for and while loops can be combined with an else clause, which is executed only when the loop is not terminated by a break statement

words = ['sky', 'apple', 'rhytm', 'fly', 'orange']

for word in words:
    for letter in word:
        # in kelimesi Türkçe'deki "-in içinde var mı?" demektir.
        if letter.lower() in 'aeiou':
            print(f"'{word} contains the vowel '{letter}' ")
            break
    else:
        print(f"'{word}' has no vowels")
```
